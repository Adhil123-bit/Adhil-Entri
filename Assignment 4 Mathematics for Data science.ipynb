```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('winequality-red.csv', sep=';')
```

```python
df.head()
```

```python
#Load the dataset and handle any missing data by replacing null values with the mean vlue of the respective column.
```

```python
# Replace missing values with the mean of their respective columns
df = df.fillna(df.mean(numeric_only=True))

# Check to confirm that missing values are handled
print(df.isnull().sum())
```

```python
#2. Extract the following columns as vectors: alcohol,citric acid.
```

```python
# Extract alcohol column as a vector
alcohol_vector = df['alcohol']

# Extract citric acid column as a vector
citric_acid_vector = df['citric acid']

# Display the vectors
print("Alcohol Vector:")
print(alcohol_vector)

print("Citric Acid Vector:")
print(citric_acid_vector)
```

###3. Select two features (e.g., alcohol and density) from the dataset and calculate the covariance matrix using np.cov(X.T), where X is the feature matrix consisting of the selected columns.(Score 2)

```python
# Select the two columns
features = ['alcohol', 'density']

# Create the feature matrix X
X = df[features].values

# Check the shape of X (should be: number of rows, 2 features)
print("Shape of X:", X.shape)

print(X)
```

```python
# Calculate the covariance matrix
cov_matrix = np.cov(X.T)

print("Covariance Matrix:")
print(cov_matrix)
```

Perform eigen decomposition on the covariance matrix you computed in question 3.
Identify and interpret the results:Identify the top 2 eigenvalues of the covariance
matrix,Identify the corresponding eigenvectors.(Score 2)

```python
# Perform eigen decomposition
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)

print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```

Which wine quality is most common in the dataset? How can you interpret the
distribution of wine quality scores?

```python
#Count the frequency of each quality score

quality_counts = df['quality'].value_counts().sort_index()

print("Distribution of Wine Quality:\n")
print(quality_counts)
```

```python
#Find the most common wine quality (the mode)

most_common = df['quality'].mode()[0]

print(f"\nThe most common wine quality is: {most_common}")
```
