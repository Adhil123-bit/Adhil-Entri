```python
# ==========================================
# BANKING SYSTEM PROGRAM
# ==========================================

# Database using Dictionaries and Nested Data Structures
accounts = {
    "1001": {"pin": "1234", "name": "Alice", "balance": 500.0},
    "1002": {"pin": "5678", "name": "Bob", "balance": 1000.0}
}

# Variable to track the currently logged-in account
logged_in_account = None
```

==========================================
# 1. ACCOUNT LOGIN CODE
# =========================================
|

```python
def login():
    global logged_in_account
    print("\n--- Account Login ---")

    # Input functions
    acc_num = input("Enter Account Number: ").strip()
    pin = input("Enter PIN: ").strip()

    # Conditional statement to validate user credentials
    if acc_num in accounts and accounts[acc_num]["pin"] == pin:
        logged_in_account = acc_num
        print(f"Login successful! Welcome, {accounts[acc_num]['name']}.")
        return True
    else:
        # Error message for invalid login
        print("Error: Invalid Account Number or PIN. Please try again.")
        return False
```

2. AMOUNT DEPOSIT CODE

```python
def deposit():
    if not logged_in_account:
        print("Error: Please log in first.")
        return

    print("\n--- Deposit Amount ---")
    # Exception handling (try-except)
    try:
        # Input function and type casting to float
        amount = float(input("Enter amount to deposit: "))

        # Operators and conditional logic
        if amount <= 0:
            print("Error: Deposit amount must be greater than zero.")
        else:
            # Update balance in dictionary
            accounts[logged_in_account]["balance"] += amount
            current_balance = accounts[logged_in_account]["balance"]
            print(f"Successfully deposited ${amount:.2f}.")
            print(f"Current Balance: ${current_balance:.2f}")

    except ValueError:
        print("Error: Invalid input. Please enter a valid numerical amount.")
```

3. AMOUNT WITHDRAWAL CODE


```python
def withdraw():
    if not logged_in_account:
        print("Error: Please log in first.")
        return

    print("\n--- Withdraw Amount ---")
    # Exception handling (try-except)
    try:
        # Input function and type casting to float
        amount = float(input("Enter amount to withdraw: "))
        current_balance = accounts[logged_in_account]["balance"]

        # Operators and conditional logic
        if amount <= 0:
            print("Error: Withdrawal amount must be greater than zero.")
        elif amount > current_balance:
            print(f"Error: Insufficient funds. Available balance is ${current_balance:.2f}.")
        else:
            # Update balance in dictionary
            accounts[logged_in_account]["balance"] -= amount
            remaining_balance = accounts[logged_in_account]["balance"]
            print(f"Successfully withdrew ${amount:.2f}.")
            print(f"Remaining Balance: ${remaining_balance:.2f}")

    except ValueError:
        print("Error: Invalid input. Please enter a valid numerical amount.")
```

MAIN APPLICATION PROGRAM

```python
def main():
    print("Welcome to the Python Banking System")

    # User authentication loop
    while not logged_in_account:
        if not login():
            retry = input("Would you like to try logging in again? (y/n): ").strip().lower()
            if retry != 'y':
                print("Exiting application. Goodbye!")
                return

    # Operations loop after successful login
    while True:
        print("\n=== Banking Operations ===")
        print("1. Deposit Money")
        print("2. Withdraw Money")
        print("3. Check Balance")
        print("4. Logout & Exit")

        choice = input("Select an option (1-4): ").strip()

        if choice == "1":
            deposit()
        elif choice == "2":
            withdraw()
        elif choice == "3":
            bal = accounts[logged_in_account]["balance"]
            print(f"\nYour current balance is: ${bal:.2f}")
        elif choice == "4":
            print(f"Logged out of account {logged_in_account}. Thank you for using our service!")
            break
        else:
            print("Invalid option. Please choose between 1 and 4.")


if __name__ == "__main__":
    main()
```
