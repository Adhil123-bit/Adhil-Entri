```python
import seaborn as sns
import matplotlib.pyplot as plt
```

#1. Countplot: Passenger Count by Embarked Port
● Create a countplot to display the number of passengers who boarded from each
port (embarked), separated by sex.

```python
df=sns.load_dataset('titanic')
```

```python
sns.countplot(data=df, x='embarked', hue='sex')


# Add title and labels
plt.title('Passenger Count by Embarked Port and Sex')
plt.xlabel('Embarked Port')
plt.ylabel('Number of Passengers')

plt.show()
```

In the above count plot we can see that most of the passengers has embarked from port S


# 2. Swarmplot: Age Distribution across Classes
● Create a swarmplot to visualize the distribution of age across passenger classes
(pclass), categorized by sex.

```python
# Create a swarmplot
sns.swarmplot(data=df, x='pclass', y='age', hue='sex')

# Add title and labels
plt.title('Age Distribution Across Passenger Classes')
plt.xlabel('Passenger Class')
plt.ylabel('Age')

plt.show()
```

From the above swarmplot we can understand that most of the passengers from 1st class,2nd class and 3rd class belongs to the age group of 20 to 40.

#3. Pie Chart: Gender Distribution
● Create a pie chart using Matplotlib to show the percentage of male vs. female
passengers. - Add percentage labels using autopct - Use explode for one slice to
highlight it.

```python
# Count male and female passengers
gender_count = df['sex'].value_counts()

# Create a pie chart
plt.pie(gender_count, labels=gender_count.index,
        autopct='%1.1f%%', explode=(0.1, 0))

plt.title('Gender Distribution of Titanic Passengers')

plt.show()
```

From the above pie chart we can understand that most of the passengers in the titanic were men with a percentage of 64.8

#4. Heatmap: Correlation Matrix
● Create a heatmap to show correlations between age, fare, pclass, sibsp, parch,
and survived. (Use seaborn.heatmap with annotations)

```python
columns = ['age', 'fare', 'pclass', 'sibsp', 'parch', 'survived']
subset_df = df[columns]

# 3. Calculate the correlation matrix
corr_matrix = subset_df.corr()

# 4. Set up the drawing canvas size
plt.figure(figsize=(8, 6))

# 5. Create the heatmap with annotations
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)

# 6. Add a title and show the plot
plt.title('Correlation Matrix Heatmap (Titanic Dataset)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

#5. Violin Plot: Fare by Class and Gender
● Create a violin plot to visualize fare distribution across classes, separated by
gender.

```python
plt.figure(figsize=(8, 5))

# 3. Create the violin plot
sns.violinplot(data=df, x='pclass', y='fare', hue='sex', split=True, palette='Set2')

# 4. Add titles and labels
plt.title('Fare Distribution Across Classes by Gender', fontsize=14, fontweight='bold')
plt.xlabel('Passenger Class (Pclass)')
plt.ylabel('Fare ($)')

# 5. Display the plot
plt.tight_layout()
plt.show()
```

#6. Scatter Plot: Age vs Fare
● Create a scatter plot to explore the relationship between age and fare, colored by
survival status.

```python
plt.figure(figsize=(9, 6))

# 3. Create the scatter plot
sns.scatterplot(
    data=df,
    x='age',
    y='fare',
    hue='survived',
    palette='coolwarm',
    alpha=0.7
)

# 4. Add titles and labels
plt.title('Age vs. Fare (Colored by Survival Status)', fontsize=14, fontweight='bold')
plt.xlabel('Age')
plt.ylabel('Fare ($)')

# 5. Display the plot
plt.tight_layout()
plt.show()
```
