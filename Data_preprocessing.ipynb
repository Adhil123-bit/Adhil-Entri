## 2. Data Cleaning: (Score : 2)
● Find the missing and inappropriate values, treat them appropriately.
● Remove all duplicate rows.
● Find the outliers.
● Replace the value 0 in age as NaN Treat the null values in all columns using any
measures(removing/ replace the values with mean/median/mode) .

```python
import numpy as np
```

```python
print("--- Original Dataset Shape ---")
print(df.shape)
```

```python
# --- STEP A: Remove Duplicate Rows ---
# Drop exact duplicate rows and keep the first occurrence
df = df.drop_duplicates()
print(f"\nShape after removing duplicates: {df.shape}")
```

```python
# --- STEP B: Find and Treat Inappropriate Values (Age = 0) ---
# Replace 0 values in the 'Age' column with NaN (Missing)
df['Age'] = df['Age'].replace(0, np.nan)
```

```python
# --- STEP C: Check Missing Values Before Treatment ---
print("\n--- Missing Values Before Treatment ---")
print(df.isnull().sum())
```

```python
# --- STEP D: Treat Null Values ---
# 1. Fill missing numerical values (like Age and Salary) with the median or mean
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Salary'] = df['Salary'].fillna(df['Salary'].median())

# 2. Fill missing categorical values (like Company and Place) with the mode (most frequent value)
df['Company'] = df['Company'].fillna(df['Company'].mode()[0])
df['Place'] = df['Place'].fillna(df['Place'].mode()[0])

print("\n--- Missing Values After Treatment ---")
print(df.isnull().sum())


# --- STEP E: Find Outliers (Using the Interquartile Range / IQR Method) ---
# We will check for outliers in numerical columns like 'Salary' and 'Age'

print("\n--- Outlier Detection (IQR Method) ---")
for col in ['Salary', 'Age']:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Filter outliners
    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"Number of outliers in '{col}': {len(outliers)}")
```

#4. Data Encoding: (Score : 2)
● Convert categorical variables into numerical representations using techniques
such as one-hot encoding, label encoding, making them suitable for analysis by
machine learning algorithms.

```python
from sklearn.preprocessing import LabelEncoder
```

```python
print("--- Original Categorical Columns ---")
print(df[['Company', 'Place']].head())
```

```python
# --- TECHNIQUE 1: One-Hot Encoding (using Pandas get_dummies) ---
# Best for categorical variables with no inherent order (like Company names or Cities/Places)
# It creates a new binary column (0 or 1) for each unique category.

df_onehot = pd.get_dummies(df, columns=['Company'], prefix='Company')

print("\n--- After One-Hot Encoding (Company Column) ---")
print(df_onehot[['Company_TCS', 'Company_Infosys', 'Company_CTS']].head())
```

```python
# --- TECHNIQUE 2: Label Encoding (using scikit-learn LabelEncoder) ---
# Best for ordinal categories or simple binary/categorical columns
# It assigns a unique integer to each category (e.g., 0, 1, 2, 3...)

label_encoder = LabelEncoder()

# We will apply Label Encoding to the 'Place' column
# Note: First, we fill any missing values with a string placeholder so LabelEncoder doesn't fail
df['Place'] = df['Place'].fillna('Unknown')
df['Place_Encoded'] = label_encoder.fit_transform(df['Place'])

print("\n--- After Label Encoding (Place Column) ---")
print(df[['Place', 'Place_Encoded']].head(10))
```

#5. Feature Scaling: (Score : 2)
● After the process of encoding, perform the scaling of the features using
standardscaler and minmaxscaler.

```python
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python
# 2. Select the numerical features we want to scale (e.g., Age and Salary)
# First, fill missing values temporarily with the median so scaling works smoothly
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
```

```python
features_to_scale = ['Age', 'Salary']
data_subset = df[features_to_scale]

print("--- Original Data (Before Scaling) ---")
print(data_subset.head())
```

```python
# --- TECHNIQUE 1: StandardScaler (Z-score Normalization) ---
# Centers the data around a mean of 0 and a standard deviation of 1.
# Formula: (x - mean) / standard_deviation

scaler_standard = StandardScaler()
scaled_standard_array = scaler_standard.fit_transform(data_subset)

# Convert the resulting numpy array back into a readable pandas DataFrame
df_standard = pd.DataFrame(scaled_standard_array, columns=['Age_StandardScaled', 'Salary_StandardScaled'])

print("\n--- After StandardScaler ---")
print(df_standard.head())
```

```python
# --- TECHNIQUE 2: MinMaxScaler (Normalization) ---
# Scales all values to a fixed range between 0 and 1.
# Formula: (x - min) / (max - min)

scaler_minmax = MinMaxScaler()
scaled_minmax_array = scaler_minmax.fit_transform(data_subset)

# Convert the resulting numpy array back into a readable pandas DataFrame
df_minmax = pd.DataFrame(scaled_minmax_array, columns=['Age_MinMaxScaled', 'Salary_MinMaxScaled'])

print("\n--- After MinMaxScaler ---")
print(df_minmax.head())
```
